```python
import pandas as pd
```

```python
res = pd.read_csv(
    "/workspace/experiments/01162026_automatedtesting/outputs/phenotype_similarity/outputs_presentation/results.csv"
)
res
```

```python
diff_phenotypes = res["transcript_cluster_1"] != res["transcript_cluster_2"]
no_unknown_1 = res["transcript_cluster_1"] != "unknown"
no_unknown_2 = res["transcript_cluster_2"] != "unknown"
with_enough_obs_1 = res["n_obs_1"] >= 3
with_enough_obs_2 = res["n_obs_2"] >= 3

n_pairs_total = res.shape[0]
n_pairs_with_different_phenotypes = (diff_phenotypes).sum()
n_pairs_with_different_phenotypes_no_control = (diff_phenotypes & no_unknown_1 & no_unknown_2).sum()
n_pairs_with_different_phenotypes_no_control_good_cells = (
    diff_phenotypes & no_unknown_1 & no_unknown_2 & with_enough_obs_1 & with_enough_obs_2
).sum()

print("n_pairs_total", n_pairs_total)
print("n_pairs_with_different_phenotypes", n_pairs_with_different_phenotypes)
print("n_pairs_with_different_phenotypes_no_control", n_pairs_with_different_phenotypes_no_control)
print(
    "n_pairs_with_different_phenotypes_no_control_good_cells",
    n_pairs_with_different_phenotypes_no_control_good_cells,
)
```
