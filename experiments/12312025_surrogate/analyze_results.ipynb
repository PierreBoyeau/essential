```python
import pandas as pd
import glob
import plotnine as gg
import numpy as np
```

```python
all_results = []
for result_file in glob.glob("results_spacer/*"):
    result = pd.read_csv(result_file)
    all_results.append(result)
all_results = pd.concat(all_results)
```

```python
all_results_pca = all_results.query("arg_use_pca == True")
all_results_expression = all_results.query("arg_use_pca == False")
```

```python
(
    gg.ggplot(all_results_expression, gg.aes(x="factor(n_labeled)", y="err", fill="approach"))
    # + gg.facet_grid("arg_model_type", "arg_fitness_cols")
    # + gg.stat_summary(geom="bar", fun_y=np.mean, position="dodge", width=0.5)
    + gg.geom_boxplot(gg.aes(fill="approach"))
    + gg.theme_minimal()
    + gg.theme(
        figure_size=(8, 4),
    )
    + gg.labs(
        y="MSE ($D_U$)",
        x="# of labeled genes",
    )
)
```

```python
(
    gg.ggplot(
        all_results_expression, gg.aes(x="factor(n_labeled)", y="corr_q0.90", fill="approach")
    )
    # + gg.facet_grid("arg_model_type", "arg_fitness_cols")
    + gg.stat_summary(geom="bar", fun_y=np.mean, position="dodge", width=0.5)
    + gg.theme_minimal()
    + gg.theme(
        figure_size=(8, 4),
    )
    + gg.labs(
        y="Correlation with GT ($D_U$)",
        x="# of labeled genes",
    )
)
```
