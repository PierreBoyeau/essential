### Notation

| Symbol                   | Meaning                                      |
| ------------------------ | -------------------------------------------- |
| $[A], [B], [C], [D]$     | Metabolite concentrations                    |
| $[g_i]_T$                | Total concentration of enzyme $g_i$          |
| $[g_i]$                  | Free (unbound) concentration of enzyme $g_i$ |
| $[Ag_1], [Bg_2], [Cg_3]$ | Enzyme-substrate complex concentrations      |

The pathway and elementary mechanism for each step:

$$A \xrightarrow{g_1} B \xrightarrow{g_2} C \xrightarrow{g_3} D$$

$$S_i + g_i \underset{k_{-1}^i}{\stackrel{k_1^i}{\rightleftharpoons}} S_ig_i \xrightarrow{k_2^i} P_i + g_i$$

where $S_i$, $P_i$ denote the substrate and product of reaction $i$, and $S_ig_i$ is the corresponding complex.

---

### Assumptions

**A0 — Closed endpoints.** A is supplied at constant rate $\alpha$; D is consumed at first-order rate $\gamma[D]$. Both are absorbed directly into the ODEs.

**A1 — Mass action kinetics.** Each elementary step proceeds at a rate proportional to the product of its reactant concentrations.

**A2 — Enzyme conservation.** Each enzyme exists in exactly two states — free or bound. Their sum is the total enzyme concentration, which is fixed (or a known function of time):

$$[g_i]_T = [g_i] + [S_ig_i]$$

This is a conservation identity that holds at every instant, including when $[g_i]_T$ is time-varying. It reduces the number of unknowns by one per reaction.

**A3 — Quasi-steady-state (QSS).** Each complex $S_ig_i$ equilibrates on a timescale of seconds, whereas metabolite concentrations and $[g_2]_T$ change on a timescale of hours. Therefore at every instant:

$$\frac{d[S_ig_i]}{dt} \approx 0$$

For $g_2$ specifically: even though $[g_2]_T$ drifts slowly in time, the drift is slow enough that $[Bg_2]$ re-equilibrates quasi-instantaneously to the current value of $[g_2]_T$. QSS remains valid.

**A4 — Substrate excess.** $[S_i] \gg [g_i]_T$, so the fraction of substrate sequestered in the complex is negligible.

**A5 — Irreversible catalysis.** The step $S_ig_i \to P_i + g_i$ is irreversible; product does not rebind the enzyme.

**A6 — Homogeneous, isothermal, constant-volume system.** All species are well-mixed; rate constants are fixed.

---

model for $[g_2]_T$

The proper equation includes a residual synthesis rate $\beta > 0$, balanced against the same loss term:

$$\frac{d[g_2]_T}{dt} = \beta - \mu\,[g_2]_T$$

This is a linear ODE with steady state:

$$[g_2]_T^* = \frac{\beta}{\mu}$$

which is the **new lower plateau** the enzyme decays toward. Solving with initial condition $[g_2]_T(0) = [g_2]_0$:

$$\boxed{[g_2]_T(t) = [g_2]_T^* + \bigl([g_2]_0 - [g_2]_T^*\bigr)\,e^{-\mu t}}$$

As $t \to \infty$, $[g_2]_T(t) \to [g_2]_T^* = \beta/\mu > 0$, which is the correct knockdown behaviour. The two limiting cases are:
- **No intervention:** $\beta = \beta_0$ (original synthesis rate), so $[g_2]_T^* = \beta_0/\mu = [g_2]_0$ — no change.
- **Knockout:** $\beta = 0$, so $[g_2]_T^* = 0$ and we recover the pure exponential decay from before.
- **Knockdown:** $0 < \beta < \beta_0$, so $0 < [g_2]_T^* < [g_2]_0$ — the enzyme settles at a reduced but nonzero level.


---

### Derivation of Michaelis-Menten rates

#### Reaction 1 — $g_1 : A \to B$, constant $[g_1]_T$

By **A1**, the mass-action ODE for the complex $[Ag_1]$ is:

$$\frac{d[Ag_1]}{dt} = k_1^1[A][g_1] - k_{-1}^1[Ag_1] - k_2^1[Ag_1]$$

Apply **A3** (QSS), setting the left-hand side to zero:

$$k_1^1[A][g_1] = \bigl(k_{-1}^1 + k_2^1\bigr)[Ag_1]$$

Substitute $[g_1] = [g_1]_T - [Ag_1]$ from **A2**:

$$k_1^1[A]\bigl([g_1]_T - [Ag_1]\bigr) = \bigl(k_{-1}^1 + k_2^1\bigr)[Ag_1]$$

Expand, collect $[Ag_1]$, and divide numerator and denominator by $k_1^1$:

$$[Ag_1] = \frac{[g_1]_T\,[A]}{K_m^1 + [A]}, \qquad K_m^1 = \frac{k_{-1}^1 + k_2^1}{k_1^1}$$

By **A1** and **A5**, the rate of product formation is:

$$v_1 = k_2^1[Ag_1] = \frac{V_{\max}^1\,[A]}{K_m^1 + [A]}, \qquad V_{\max}^1 = k_2^1[g_1]_T \quad \text{(constant)}$$

---

#### Reaction 2 — $g_2 : B \to C$, time-varying $[g_2]_T(t)$

By **A1**, the mass-action ODE for $[Bg_2]$ is:

$$\frac{d[Bg_2]}{dt} = k_1^2[B][g_2] - k_{-1}^2[Bg_2] - k_2^2[Bg_2]$$

Apply **A3** (QSS is still valid by timescale separation, even though $[g_2]_T$ drifts):

$$k_1^2[B][g_2] = \bigl(k_{-1}^2 + k_2^2\bigr)[Bg_2]$$

Substitute $[g_2] = [g_2]_T(t) - [Bg_2]$ from **A2**, evaluated at the current value of $[g_2]_T(t)$:

$$k_1^2[B]\bigl([g_2]_T(t) - [Bg_2]\bigr) = \bigl(k_{-1}^2 + k_2^2\bigr)[Bg_2]$$

Solving identically as before:
$$[Bg_2] = \frac{[g_2]_T(t)\,[B]}{K_m^2 + [B]}, \qquad K_m^2 = \frac{k_{-1}^2 + k_2^2}{k_1^2}$$

$K_m^2$ depends only on rate constants (**A1**, **A6**) — it is unaffected by the knockdown. The rate is:

$$v_2(t) = k_2^2[Bg_2] = \frac{V_{\max}^2(t)\,[B]}{K_m^2 + [B]}, \qquad V_{\max}^2(t) = k_2^2\left[\frac{\beta}{\mu} + \left([g_2]_0 - \frac{\beta}{\mu}\right)e^{-\mu t}\right]$$

---

#### Reaction 3 — $g_3 : C \to D$, constant $[g_3]_T$

Identical derivation to reaction 1:

$$v_3 = \frac{V_{\max}^3\,[C]}{K_m^3 + [C]}, \qquad V_{\max}^3 = k_2^3[g_3]_T \quad \text{(constant)}, \qquad K_m^3 = \frac{k_{-1}^3 + k_2^3}{k_1^3}$$

---

### Complete ODE system

$$\boxed{\begin{aligned}
\frac{d[A]}{dt} &= \alpha - \frac{V_{\max}^1[A]}{K_m^1+[A]} \\[10pt]
\frac{d[B]}{dt} &= \frac{V_{\max}^1[A]}{K_m^1+[A]} - \frac{V_{\max}^2(t)[B]}{K_m^2+[B]} \\[10pt]
\frac{d[C]}{dt} &= \frac{V_{\max}^2(t)[B]}{K_m^2+[B]} - \frac{V_{\max}^3[C]}{K_m^3+[C]} \\[10pt]
\frac{d[D]}{dt} &= \frac{V_{\max}^3[C]}{K_m^3+[C]} - \gamma[D]
\end{aligned}}$$


**When does the system have a steady-state solution?**
Necessary conditions:

$$\begin{align*} \begin{cases} \alpha &= \dfrac{V_{\max}^1 [A]_\infty}{K_m^1 + [A]_\infty} \\ \dfrac{V_{\max}^1 [A]_\infty}{K_m^1 + [A]_\infty} &= \dfrac{V_{\max}^{2,\infty} [B]_\infty}{K_m^2 + [B]_\infty} \\ \dfrac{V_{\max}^{2,\infty} [B]_\infty}{K_m^2 + [B]_\infty} &= \dfrac{V_{\max}^3 [C]_\infty}{K_m^3 + [C]_\infty} \\ \dfrac{V_{\max}^3 [C]_\infty}{K_m^3 + [C]_\infty} &= \gamma [D]_\infty \end{cases} \end{align*}$$

Note that these conditions also imply that all reactions fluxes are equal at equilibrium: $v_1, \lim_{t\rightarrow \infty} v_2(t) = v_3$


$$\begin{align*} 
\begin{cases} 
[A]_\infty &= \dfrac{\alpha K_m^1}{V_{\max}^1 - \alpha} \\ 
[B]_\infty &= \dfrac{\alpha K_m^2}{V_{\max}^{2,\infty} - \alpha} \\
[C]_\infty &= \dfrac{\alpha K_m^3}{V_{\max}^3 - \alpha} \\
[D]_\infty &= \dfrac{\alpha}{\gamma} 
\end{cases} 
\end{align*}$$



```python
from simulator import simulate
import matplotlib.pyplot as plt
import numpy as np
import plotnine as gg
import pandas as pd

from essential.utils import PLOTNINE_DEFAULT_THEME_2


plt.rcParams["svg.fonttype"] = "none"


def create_single_df(M, t, M_name):
    return pd.DataFrame(
        {
            "concentration": M,
            "time": t,
            "metabolite": M_name,
        }
    )

def create_df(A, B, C, D, g2, t):
    return pd.concat([create_single_df(A, t, "A"), create_single_df(B, t, "B"), create_single_df(C, t, "C"), create_single_df(D, t, "D"), create_single_df(g2, t, "g2")])
```

```python
import numpy as np
import pandas as pd
import plotnine as gg

# Data
generations = np.linspace(0, 5, 500)
betas = {"0.50": 0.50, "0.10": 0.10}

records = []
for label, plateau in betas.items():
    conc = plateau + (1 - plateau) * np.exp(-np.log(2) * generations)
    for t, c in zip(generations, conc):
        records.append({"generation": t, "concentration": c, "knockdown": label})

df = pd.DataFrame(records)

# Asymptote lines
hlines = pd.DataFrame({"yintercept": [0.50, 0.10], "knockdown": ["0.50", "0.10"]})

fig = (
    gg.ggplot(df, gg.aes(x="generation", y="concentration", color="knockdown"))
    + gg.geom_line(size=0.8)
    + gg.geom_hline(gg.aes(yintercept="yintercept", color="knockdown"),
                    data=hlines, linetype="dashed", size=0.5, alpha=0.6)
    + gg.annotate("text", x=5.1, y=0.50, label="$\\beta/\\mu = 0.50$",
                  ha="left", va="center", size=7, color="#E69F00")
    + gg.annotate("text", x=5.1, y=0.10, label="$\\beta/\\mu = 0.10$",
                  ha="left", va="center", size=7, color="#0072B2")
    + gg.scale_x_continuous(breaks=[0, 1, 2, 3, 4, 5], expand=(0, 0))
    + gg.scale_y_continuous(limits=(0, 1.05), breaks=[0, 1], expand=(0,0),
                             labels=["0", r"$[g]_0$"])
    + gg.scale_color_manual(values={"0.50": "#E69F00", "0.10": "#0072B2"})
    + gg.labs(x="Time (cell generations)", y=r"$[g]_T$")
    + gg.theme_minimal()
    + gg.theme(figure_size=(3.5, 2.5), legend_position="none")
    + PLOTNINE_DEFAULT_THEME_2
)
fig.save("knockdown_illustration.svg")
fig
```

```python
# Shared
alpha  = 1.0
gamma  = 1.0
KM1 = KM2 = KM3 = 1.0
k2_1 = 1.2
k2_2   = 3.0   # Vmax1 = 2.0 > alpha
k2_3   = 1.2   # Vmax3 = 2.0 > alpha
g1T0   = 1.0
g2T0   = 1.0
g3T0   = 1.0
mu     = np.log(2)
Tmax   = 50
n_time_points = 1000
```

```python
beta = mu * 0.2 * g2T0
print("enzyme level at steady state", beta / mu)
vmax2 = k2_2 * beta / mu
print("vmax2", vmax2)
print("alpha", alpha)


A, B, C, D, g2, t = simulate(
    alpha, gamma, KM1, KM2, KM3, k2_1, k2_2, k2_3, g1T0, g2T0, g3T0, mu, beta, Tmax, n_time_points
)

RENAMER = {"A": "[A]", "B": "[B]", "C": "[C]", "D": "[D]"}

all_df = create_df(A, B, C, D, g2, t)
metabolite_df = all_df[all_df["metabolite"].isin(["A", "B", "C", "D"])].assign(metabolite=lambda df: df["metabolite"].map(RENAMER))

kd_df = all_df[all_df["metabolite"] == "g2"]
metabolite_df1 = metabolite_df.copy()

fig1 = (
    gg.ggplot(metabolite_df, gg.aes(x="time", y="concentration", color="metabolite"))
    + gg.geom_line()
    + gg.theme_minimal()
    + PLOTNINE_DEFAULT_THEME_2
    + gg.theme(figure_size=(1.5, 1.2))
)
display(fig1)

vmax = kd_df["concentration"].max() + 0.1
fig2 = (
    gg.ggplot(kd_df, gg.aes(x="time", y="concentration"))
    + gg.geom_line()
    + gg.theme_minimal()
    + gg.scale_y_continuous(limits=(0, vmax))
    + PLOTNINE_DEFAULT_THEME_2
    + gg.theme(figure_size=(1.5, 1.2))
)
display(fig2)
fig2.save("overflow.svg")
```

```python
beta = mu * 0.4 * g2T0
print("enzyme level at steady state", beta / mu)
vmax2 = k2_2 * beta / mu
print("vmax2", vmax2)
print("alpha", alpha)


A, B, C, D, g2, t = simulate(
    alpha, gamma, KM1, KM2, KM3, k2_1, k2_2, k2_3, g1T0, g2T0, g3T0, mu, beta, Tmax, n_time_points
)
all_df = create_df(A, B, C, D, g2, t)
metabolite_df = all_df[all_df["metabolite"].isin(["A", "B", "C", "D"])].assign(metabolite=lambda df: df["metabolite"].map(RENAMER))
kd_df = all_df[all_df["metabolite"] == "g2"]

metabolite_df2 = metabolite_df.copy()


fig1 = (
    gg.ggplot(metabolite_df, gg.aes(x="time", y="concentration", color="metabolite"))
    + gg.geom_line()
    + gg.theme_minimal()
    + PLOTNINE_DEFAULT_THEME_2
)
display(fig1)

vmax = kd_df["concentration"].max() + 0.1
fig2 = (
    gg.ggplot(kd_df, gg.aes(x="time", y="concentration"))
    + gg.geom_line()
    + gg.theme_minimal()
    + gg.scale_y_continuous(limits=(0, vmax))
    + PLOTNINE_DEFAULT_THEME_2
    + gg.theme(figure_size=(2, 1.4))
)
display(fig2)
fig2.save("buffer_poor.svg")
```

```python
beta = mu * 0.6 * g2T0
print("enzyme level at steady state", beta / mu)
vmax2 = k2_2 * beta / mu
print("vmax2", vmax2)
print("alpha", alpha)


A, B, C, D, g2, t = simulate(
    alpha, gamma, KM1, KM2, KM3, k2_1, k2_2, k2_3, g1T0, g2T0, g3T0, mu, beta, Tmax, n_time_points
)
all_df = create_df(A, B, C, D, g2, t)
metabolite_df = all_df[all_df["metabolite"].isin(["A", "B", "C", "D"])].assign(metabolite=lambda df: df["metabolite"].map(RENAMER))
kd_df = all_df[all_df["metabolite"] == "g2"]

metabolite_df3 = metabolite_df.copy()


fig1 = (
    gg.ggplot(metabolite_df, gg.aes(x="time", y="concentration", color="metabolite"))
    + gg.geom_line()
    + gg.theme_minimal()
    + PLOTNINE_DEFAULT_THEME_2
)
display(fig1)

vmax = kd_df["concentration"].max() + 0.1
fig2 = (
    gg.ggplot(kd_df, gg.aes(x="time", y="concentration"))
    + gg.geom_line()
    + gg.theme_minimal()
    + gg.scale_y_continuous(limits=(0, vmax))
    + PLOTNINE_DEFAULT_THEME_2
    + gg.theme(figure_size=(2, 1.4))
)
display(fig2)
fig2.save("buffer_strong.svg")
```

```python
all_metabolites = pd.concat(
    [
        metabolite_df1.assign(regime="overflow"),
        metabolite_df2.assign(regime="buffer (poor)"),
        metabolite_df3.assign(regime="buffer (strong)"),
    ]
)

regime_order = ["overflow", "buffer (poor)", "buffer (strong)"]
all_metabolites["regime"] = pd.Categorical(
    all_metabolites["regime"], categories=regime_order, ordered=True
)
```

```python
fig1 = (
    gg.ggplot(all_metabolites, gg.aes(x="time", y="concentration", color="metabolite"))
    + gg.geom_line()
    + gg.facet_wrap("regime")
    + gg.theme_minimal()
    + PLOTNINE_DEFAULT_THEME_2
    + gg.theme(figure_size=(4, 1.5))
)
display(fig1)
fig1.save("response_comparison.svg")
```
